```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as svs
batsman = pd.read_csv("/content/sharma-kohli-ipl-2008-2026-1.csv")
batsman
plt.plot(batsman["Year"],batsman["V Kohli"],color="red")
plt.plot(batsman["Year"],batsman["RG Sharma"],color="blue")
plt.title("Virat Kohli vs Rohit Sharma — IPL Performance Analysis (2008–2026)")
plt.legend("VR")
plt.xlabel("season")
plt.ylabel("Run Scored")
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(batsman["V Kohli"], bins=8, alpha=0.6, label="Virat Kohli",color="red")
plt.hist(batsman["RG Sharma"], bins=8, alpha=0.6, label="Rohit Sharma")

plt.xlabel("Runs")
plt.ylabel("Number of Seasons")
plt.title("IPL Run Distribution: Virat Kohli vs Rohit Sharma")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
total_kohli = batsman["V Kohli"].sum()
total_rohit = batsman["RG Sharma"].sum()

runs = [total_kohli, total_rohit]
players = ["Virat Kohli", "Rohit Sharma"]

plt.pie(runs,labels=players,autopct="%1.1f%%",colors=["Red","Blue"])

plt.title("IPL Run Share: Virat Kohli vs Rohit Sharma")
plt.show()
```
